### New DF

```python
# Model
COAD_Full = COAD_Model[['ModelID', 'PatientID', 'CellLineName', 'PrimaryOrMetastasis',
                        'SampleCollectionSite', 'Age', 'Sex', 'GrowthPattern']].copy()

COAD_Full.head(2)
```

```python
# Expression
drop_col = ['Unnamed: 0', 'IsDefaultEntryForMC', 'IsDefaultEntryForModel']
COAD_Expression_2 = COAD_Expression.drop(columns=drop_col).copy()
COAD_Expression_2.head(2)
```

```python
# CRISPRGeneEffect
print(COAD_CRISPRGeneEffect.shape)
COAD_CRISPRGeneEffect_2 = COAD_CRISPRGeneEffect.dropna(axis=1, how='all').copy()
print(COAD_CRISPRGeneEffect_2.shape)

COAD_CRISPRGeneEffect_2.head(2)
```

```python
# Mutation
COAD_Mutation_2 = COAD_Mutation[['ModelID', 'SequencingID', 'ModelConditionID', 'ProteinChange', 'HugoSymbol', 'VariantType', 'Hotspot']].copy()
COAD_Mutation_2.head(2)
```

```python
# Clear data without CRISPRGeneEffect

print("Before filtering COAD_Full:", COAD_Full.shape)
COAD_Full = COAD_Full[COAD_Full['ModelID'].isin(COAD_CRISPRGeneEffect_2['ModelID'].unique())].copy()
print("After filtering COAD_Full:", COAD_Full.shape)
```

### For each ModelID

```python
limit = 2
for idx, row in COAD_Full.iterrows():
    if idx >= limit:
        break

    display(row.to_frame().T)

    def clear_df(df, model_id_col, drop_col_names):
        df = df[df[model_id_col] == row[model_id_col]]
        df = df.drop(columns=drop_col_names)
        df = df.dropna(axis=1, how='all')
        return df
    

    exp = clear_df(COAD_Expression_2, 'ModelID', ['ModelID', 'ModelConditionID', 'SequencingID'])
    display("Expression", exp.head(2))
   
    mut = clear_df(COAD_Mutation_2, 'ModelID', ['ModelID', 'ModelConditionID', 'SequencingID'])
    display("Mutation", mut.head(2))

    crispr = clear_df(COAD_CRISPRGeneEffect_2, 'ModelID', ['ModelID'])
    display("CRISPRGeneEffect", crispr.head(2))
```
